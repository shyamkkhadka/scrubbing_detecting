```python
# Code to find always-on protected prefixes of each scrubbers in 30 days window
import os
import pandas as pd
from functools import reduce

# List of scrubbers (folders you have)
scrubbers = ["13335", "32787", "19905", "19551", "198949"]  # <-- update with your actual scrubbers

for scrubber in scrubbers:
    folder = f"../data/always-on/as{scrubber}"
    
    # Collect sets of prefixes from each day
    prefix_sets = []
    
    
    for day in range(1, 31):  # days 1 to 30
        filename = f"as_{scrubber}_transited_prefix_2025-05-{day:02d}.csv"
        filepath = os.path.join(folder, filename)
        if not os.path.exists(filepath):
            raise FileNotFoundError(f"Missing file: {filepath}")

        df = pd.read_csv(filepath, usecols=["prefix"])
        prefix_sets.append(set(df["prefix"]))

    # Intersect all sets → prefixes present in all days
    common_prefixes = reduce(set.intersection, prefix_sets)

    print(f"Total prefixes present every day (1–30) for scrubber AS{scrubber}: {len(common_prefixes)}")
```
