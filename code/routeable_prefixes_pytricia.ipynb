```python
# Store all the registered IP prefixes in pytricia tree
import pytricia
import requests

# Initialize a PyTricia trie
trie = pytricia.PyTricia()

def download_prefixes():
    urls = [
#         'https://ftp.ripe.net/ripe/stats/delegated-ripencc-latest',
        'https://ftp.apnic.net/stats/apnic/delegated-apnic-latest',
#         'https://ftp.lacnic.net/pub/stats/lacnic/delegated-lacnic-latest',
#         'https://ftp.afrinic.net/stats/afrinic/delegated-afrinic-latest',
#         'https://ftp.arin.net/pub/stats/arin/delegated-arin-extended-latest'
    ]
    
    prefixes = []
    
    for url in urls:
        response = requests.get(url)
        lines = response.text.splitlines()
        for line in lines:
            parts = line.split('|')
            if len(parts) >= 7 and parts[2] == 'ipv4' and parts[6] == 'allocated':
                ip_prefix = f"{parts[3]}/{32 - (int(parts[4])).bit_length()}"
                prefixes.append(ip_prefix)
                
    return prefixes

# Load prefixes into the trie
prefixes = download_prefixes()
for prefix in prefixes:
    trie[prefix] = "Routable"

# Example of how to check if an IP address is in the trie
ip_address = '192.0.2.1'
if ip_address in trie:
    print(f"{ip_address} is in a routable prefix")
else:
    print(f"{ip_address} is not in a routable prefix")
```
