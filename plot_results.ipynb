```python
# Prefix re-origination cases 
# For Cloudflare
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import io
import numpy as np
import datetime

# Globally update the font size
plt.rcParams.update({'font.size': 32})
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42

# Read the data into a pandas DataFrame
df = pd.read_csv("../data/prefix re-origination/as13335_may_original.csv")

# Rename columns for easier access
df.columns = [
    "Date",
    "Total Prefixes at 00:00",
    "New Prefixes on that day",
    "Prefixes with different origin next day"
]

# Convert 'Date' column to datetime objects
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")


# Convert other columns to numeric, coercing errors to NaN and then filling NaN with 0
df["Total Prefixes at 00:00"] = pd.to_numeric(df["Total Prefixes at 00:00"], errors='coerce').fillna(0)

# --- Scale the blue line values to thousands ---
df["Total Prefixes at 00:00 (x100)"] = df["Total Prefixes at 00:00"] / 100

df["New Prefixes on that day"] = pd.to_numeric(df["New Prefixes on that day"], errors='coerce').fillna(0)
df["Prefixes with different origin next day"] = pd.to_numeric(df["Prefixes with different origin next day"], errors='coerce').fillna(0)

# Plot setup
fig, ax = plt.subplots(figsize=(20, 10))

# Plot line for "Total Prefixes at 00:00"
ax.plot(df["Date"], df["Total Prefixes at 00:00 (x100)"], marker='o', color='blue', label="Originated at 00:00 (x100)", linewidth=2)

# Bars for "New Prefixes on that day" and "Prefixes with different origin next day"
bar_width = pd.Timedelta(days=0.25)   # width of each bar
bar_spacing = pd.Timedelta(days=0.15)  # increased gap

bars1 = ax.bar(
    df["Date"] - (bar_width / 2 + bar_spacing / 2),
    df["New Prefixes on that day"],
    width=bar_width,
    color='red',
    label="New originated prefixes",
    alpha=0.7
)

bars2 = ax.bar(
    df["Date"] + (bar_width / 2 + bar_spacing / 2),
    df["Prefixes with different origin next day"],
    width=bar_width,
    label = "Origin change next day",
    color='green',
    alpha=0.7
)


# --- X-axis formatting (show all dates from 1 to 31 within the data range) ---
date_form = mdates.DateFormatter("%d-%m-%Y")
ax.xaxis.set_major_formatter(date_form)

min_date = df["Date"].min()
max_date = df["Date"].max()
all_dates_in_range = pd.date_range(start=min_date.floor('D'), end=max_date.ceil('D'), freq='D')
ax.set_xticks(all_dates_in_range[::2]) # Show every 3rd day
plt.xticks(rotation=90) # Rotate labels for better readability
# --- End of x-axis formatting ---

# --- Y-axis formatting (show y-ticks) ---
y_min = 0
y_max_total = df["Total Prefixes at 00:00 (x100)"].max()
y_max_new_eod = df["New Prefixes on that day"].max()
y_max_diff_origin = df["Prefixes with different origin next day"].max()

y_max_data = max(y_max_total, y_max_new_eod, y_max_diff_origin)

# Increased padding slightly to accommodate bar labels
y_max_plot = y_max_data * 1.15

if y_max_data > 1000:
    y_tick_step = 500
elif y_max_data > 100:
    y_tick_step = 50
else:
    y_tick_step = 10

if y_tick_step == 0:
    y_ticks = [y_min, y_max_plot]
else:
    y_ticks = np.arange(y_min, y_max_plot + y_tick_step, y_tick_step)

ax.set_yticks(y_ticks)
# --- End of y-axis formatting ---

# --- Add values on top of bar plots ---
def autolabel(bars, color): # Added 'color' parameter
    """Attach a text label above each bar in *bars*, displaying its height."""
    for bar in bars:
        height = bar.get_height()
        if height > 0: # Only label if height is greater than 0
            ax.text(bar.get_x() + bar.get_width() / 2, height,
                    f'{int(height)}',
                    ha='center', va='bottom', color=color, rotation = 30) # Use the passed color

autolabel(bars1, 'red') # Pass 'green' for the first set of bars
autolabel(bars2, 'green')   # Pass 'red' for the second set of bars
# --- End of adding values ---

# General plot formatting
plt.xlabel("Date", weight = "bold")
plt.ylabel("No. of prefixes", weight = "bold")
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.legend(loc='center', bbox_to_anchor=(0.24, 0.78))
plt.tight_layout()
plt.savefig("../images/cloudflare_origination.pdf")
plt.show()
```

```python
# Plot that shows origin-change model of scrubbing for same-day and cross-day

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import matplotlib.ticker as mticker
import matplotlib.cm as cm
import matplotlib.colors as mcolors
import numpy as np

# Pick 3 distinct colors from Viridis
tab10 = cm.get_cmap("tab10")
colors = [tab10(0), tab10(1), tab10(2)]

# Globally update the font size
plt.rcParams.update({'font.size': 34})
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42

scrubber = "19905"

# Origin-change model file (same-day)
df1 = pd.read_csv("../data/prefix re-origination/as"+scrubber+"_may_duration.csv") 

# Origin-change model file (cross-day)
df2 = pd.read_csv("../data/diff_ribs/prefix re_origination/as"+scrubber+"_diff_may_reorigination_details_duration.csv") 

# Standardize column names for merging
df1 = df1.rename(columns={'Prefix':'prefix', 'StartTime':'StartTime', 'EndTime':'EndTime'})
df2 = df2.rename(columns={'prefix':'prefix', 'StartTime':'StartTime', 'EndTime':'EndTime'})

# Keep only relevant columns
df1 = df1[['prefix', 'StartTime', 'EndTime']]
df2 = df2[['prefix', 'StartTime', 'EndTime']]

# Merge both datasets
df = pd.concat([df1, df2], ignore_index=True)

# Convert StartTime and EndTime to datetime
df['StartTime'] = pd.to_datetime(df['StartTime'])
df['EndTime'] = pd.to_datetime(df['EndTime'])

# Determine scrubbing type
df['ScrubType'] = df.apply(lambda row: 'same-day' if row['StartTime'].date() == row['EndTime'].date() else 'cross-day', axis=1)

# Extract start date
df['StartDate'] = df['StartTime'].dt.date

# Aggregate counts per day and type
daily_counts = df.groupby(['StartDate', 'ScrubType']).size().unstack(fill_value=0)

# Reorder columns so cross-day is first, same-day is second (same-day appears on top)
# daily_counts = daily_counts[['same-day', 'cross-day']]

# Ensure all dates from May 1 to May 30 are included
all_dates = pd.date_range(start="2025-05-01", end="2025-05-30").date
daily_counts = daily_counts.reindex(all_dates, fill_value=0)

# Create figure and axes first
fig, ax = plt.subplots(figsize=(18,10))

# Plot stacked bar chart on the axes
daily_counts.plot(kind='bar', stacked=True, color=[colors[0], colors[1]], width=0.8, ax=ax)

# Replace x-axis labels with day of month
ax.set_xticks(range(len(daily_counts)))
ax.set_xticklabels([d.day for d in daily_counts.index])

# Labels and title
ax.set_xlabel('Day')

ax.set_ylabel('# Scrubbing-activations')
# ax.set_title('Daily Number of Prefixes with Same-day vs Cross-day Scrubbing (May 2025)')

# After plotting
ax.set_yticks(np.arange(0, daily_counts.values.max() + 2, 1))  # integer ticks from 0 to max +1
ax.set_ylim(0, daily_counts.values.max() + 1)  # adjust y-limit to fit bars

# Rotate x-ticks
plt.xticks(rotation=90)

# Legend
ax.legend()
ax.legend(
    loc="center",
    title='Scrubbing session',
    ncol=1,            # number of columns (adjust for layout)
    bbox_to_anchor=(0.23, 0.76),  # centered above the plot

)

plt.tight_layout()
plt.savefig("../images/vercara_origination.pdf")

plt.show()
```
